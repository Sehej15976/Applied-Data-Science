```python
#label encoding
def label_encode(df):
    label_encoder = LabelEncoder()
    for column in df.columns:
        if df[column].dtype == 'bool':
            df[column] = df[column].astype(int)
        elif df[column].dtype == 'object':
            df[column] = label_encoder.fit_transform(df[column])
    return df

df = label_encode(df)
```

```python
# prompt: code to take mean meadian and mode of each column in df

# Calculate the mean, median, and mode for each column
def calculate_stats(df):
    stats_df = pd.DataFrame(index=['mean', 'median', 'mode'])
    for col in df.columns:
        if pd.api.types.is_numeric_dtype(df[col]):
            stats_df[col] = [df[col].mean(), df[col].median(), df[col].mode()[0]]
    return stats_df

stats_df = calculate_stats(df)
stats_df
```

```python
# Calculate range
def range_fn(df):
    range_df = pd.DataFrame(index=['range'])
    for col in df.columns:
        if pd.api.types.is_numeric_dtype(df[col]):
            range_df[col] = [df[col].max() - df[col].min()]
    return range_df

range_df = range_fn(df)
print(range_df)
```

```python
# prompt: code for absolute deviation

# Calculate absolute deviation
def absolute_deviation(df):
    absolute_deviation_df = pd.DataFrame(index=['absolute_deviation'])
    for col in df.columns:
        if pd.api.types.is_numeric_dtype(df[col]):
          # Calculate the mean absolute deviation manually
          mean = df[col].mean()
          absolute_deviations = np.abs(df[col] - mean)

          mad = absolute_deviations.mean()
          absolute_deviation_df[col] = [mad] # Assign the calculated mad value
    return absolute_deviation_df

absolute_deviation_df = absolute_deviation(df)
absolute_deviation_df
```

```python
# prompt: code for standard deviation without using inbuild function

# Calculate standard deviation without using the built-in function
def standard_deviation_manual(df):
    std_dev_df = pd.DataFrame(index=['standard_deviation_manual'])
    for col in df.columns:
        if pd.api.types.is_numeric_dtype(df[col]):
            # Calculate the mean
            mean = df[col].mean()
            # Calculate squared differences from the mean
            squared_diffs = [(x - mean) ** 2 for x in df[col]]
            # Calculate the variance
            variance = sum(squared_diffs) / len(df[col])
            # Calculate the standard deviation
            std_dev = variance ** 0.5
            std_dev_df[col] = [std_dev]
    return std_dev_df

std_dev_manual_df = standard_deviation_manual(df)
std_dev_manual_df
```

```python
# prompt: code for quartile

# Calculate quartiles
def quartiles_fn(df):
    quartiles_df = pd.DataFrame(index=['Q1', 'Q2', 'Q3'])
    for col in df.columns:
        if pd.api.types.is_numeric_dtype(df[col]):
            quartiles = df[col].quantile([0.25, 0.5, 0.75])
            quartiles_df[col] = [quartiles[0.25], quartiles[0.5], quartiles[0.75]]
    return quartiles_df

quartiles_df = quartiles_fn(df)
quartiles_df
```

```python
#combined data, length,mean,median and standard deviation for use in functions after this block

numeric_df = df.select_dtypes(include=np.number)
combined_data = numeric_df.values.flatten()
n = len(combined_data)
# mean
mean = sum(combined_data) / n

# Sort the combined data
sorted_data = sorted(combined_data)

# Calculate the median
n = len(sorted_data)
if n % 2 == 0:
    # Even number of elements, median is the average of the middle two
    mid1 = sorted_data[n // 2 - 1]
    mid2 = sorted_data[n // 2]
    median = (mid1 + mid2) / 2
else:
    # Odd number of elements, median is the middle element
    median = sorted_data[n // 2]

#standard deviation of combined data
def combined_std_dev(mean,n,combined_data):

    # squared differences from the mean
    squared_diffs = [(x - mean) ** 2 for x in combined_data]

    # variance
    variance = sum(squared_diffs) / n

    # standard deviation
    std_dev = variance ** 0.5

    return std_dev

combined_std = combined_std_dev(mean,n,combined_data)

print(f"Combined Standard Deviation: {combined_std}")
print(f"Manually calculated median: {median}")
print(f"Manually calculated mean: {mean}")
```

```python
# prompt: code to calculate skewness of combined data using following formula
# summation {(X-MEAN)^3 / (N-1)*(STANDARD DEVIATION)^3}

def calculate_skewness(combined_data,combined_std,n,mean):
    # Calculate skewness using the formula
    summation = (np.sum(combined_data - mean)**3) / ((n - 1) * (combined_std**3))
    return summation

def classify_skewness(sv):
    if sv > 0:
        classification = "Right-Skewed "
    elif sv == 0:
        classification = "Symmetrical "
    else:
        classification = "Left-Skewed "
    return classification

skewness_value = calculate_skewness(combined_data,combined_std,n,mean)
classification = classify_skewness(skewness_value)
print(f"Skewness for all combined columns: {skewness_value}")
print(f"Classification of skewness: {classification}")
```

```python
# prompt: code for bouleys coeff of skewness using following formula for all column combined
# (q3-2q2+q1)/(q3-q1) where q is quratile

def bouleys_coeff_of_skewness(combined_data):
    q1 = np.quantile(combined_data, 0.25)
    q2 = np.quantile(combined_data, 0.5)
    q3 = np.quantile(combined_data, 0.75)

    skewness = (q3 - 2 * q2 + q1) / (q3 - q1)
    return skewness

bowley_skewness = bouleys_coeff_of_skewness(combined_data)
print(f"Bowley's Coefficient of Skewness for all combined columns: {bowley_skewness}")
```

```python
# prompt: code for karl pearson for combined data with formula 3*(mean-median)/standard deviation

def karl_pearson(combined_data,mean,combined_std,median):
    calm_pierson = 3 * (mean - median) / combined_std
    return calm_pierson

karl_pearson_value = karl_pearson(combined_data,mean,combined_std,median)
print(f"karl-pearson coefficient for combined data: {karl_pearson_value}")
```

```python
# prompt: code to calculate karl pearson using this formula (mean -  mode )/standard deviation for combind_data

def diff_karl_pearson(combined_data, mean, combined_std):
    from scipy import stats
    mode = stats.mode(combined_data)[0]
    karl_pearson_coeff = (mean - mode) / combined_std
    return karl_pearson_coeff

diff_karl_pearson_value = diff_karl_pearson(combined_data, mean, combined_std)
print(f"Karl Pearson coefficient for combined data: {diff_karl_pearson_value}")
```

```python
# prompt: code for kurtosis using formula for all columns combined not each column
# THIS IS FORMULA FOR KURTOSIS
# SUMMATION {(X-MEAN)^4 / (N-1)*(STANDARD DEVIATION)^4 }

def calculate_kurtosis(combined_data,combined_std,n,mean):
    # Calculate kurtosis using the formula
    summation = (np.sum(combined_data - mean)**4) / ((n - 1) * (combined_std**4))
    return summation

# classification of kurtosis
def classify_kurtosis(dk):
    if dk > 3:
        classification = "Leptokurtic "
    elif dk == 3:
        classification = "Mesokurtic "
    else:
        classification = "Platykurtic "
    return classification


kurtosis_value = calculate_kurtosis(combined_data,combined_std,n,mean)
classification = classify_kurtosis(kurtosis_value)
print(f"Kurtosis for all combined columns: {kurtosis_value}")
print(f"Classification of kurtosis: {classification}")
```
