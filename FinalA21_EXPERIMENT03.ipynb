```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.metrics import accuracy_score
import numpy as np
from sklearn.svm import SVC
```

```python
# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/ML/WA_Fn-UseC_-HR-Employee-Attrition.csv')
```

```python
def label_encode(df):
    label_encoder = LabelEncoder()
    for column in df.columns:
        if df[column].dtype == 'bool':
            df[column] = df[column].astype(int)
        elif df[column].dtype == 'object':
            df[column] = label_encoder.fit_transform(df[column])
    return df

df = label_encode(df)
```

```python

# Remove zero variance features
non_zero_variance_features = df.loc[:, df.var() > 0]
X = non_zero_variance_features.drop(columns=['Attrition'])
y = df['Attrition']

# Scale numeric features
numeric_columns = X.select_dtypes(include=[np.number]).columns
scaler = MinMaxScaler()
X[numeric_columns] = scaler.fit_transform(X[numeric_columns])

# Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train and evaluate the model before feature selection
model = LogisticRegression()
model.fit(X_train, y_train)
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)
train_accuracy_before = accuracy_score(y_train, y_train_pred)
test_accuracy_before = accuracy_score(y_test, y_test_pred)

# Print training and testing accuracy before feature selection
print(f"Training Accuracy before feature selection: {train_accuracy_before:.4f}")
print(f"Testing Accuracy before feature selection: {test_accuracy_before:.4f}")
print("*************************************************************")

# K-Fold Cross Validation for Logistic Regression before feature selection
kfold = KFold(n_splits=10, shuffle=True, random_state=42)  # Added random_state for reproducibility
accuracy_scores_lr_before = cross_val_score(model, X, y, cv=kfold, scoring='accuracy')  # Replaced model_lr with model
for i, accuracy in enumerate(accuracy_scores_lr_before):
    print(f'Logistic Regression - Accuracy for fold {i+1} before feature selection: {accuracy:.4f}')
avg_accuracy_lr_before = accuracy_scores_lr_before.mean()
print(f"Logistic Regression - Average accuracy across all folds before feature selection: {avg_accuracy_lr_before:.4f}")
print("*************************************************************")


# Train and evaluate SVM model before feature selection
model_svm = SVC()
model_svm.fit(X_train, y_train)
y_train_pred_svm = model_svm.predict(X_train)
y_test_pred_svm = model_svm.predict(X_test)
train_accuracy_svm_before = accuracy_score(y_train, y_train_pred_svm)
test_accuracy_svm_before = accuracy_score(y_test, y_test_pred_svm)

# Print training and testing accuracy for SVM before feature selection
print(f"SVM - Training Accuracy before feature selection: {train_accuracy_svm_before:.4f}")
print(f"SVM - Testing Accuracy before feature selection: {test_accuracy_svm_before:.4f}")
print("*************************************************************")



# K-Fold Cross Validation for SVM before feature selection
accuracy_scores_svm_before = cross_val_score(model_svm, X, y, cv=kfold, scoring='accuracy')
for i, accuracy in enumerate(accuracy_scores_svm_before):
    print(f'SVM - Accuracy for fold {i+1} before feature selection: {accuracy:.4f}')
avg_accuracy_svm_before = accuracy_scores_svm_before.mean()
print(f"SVM - Average accuracy across all folds before feature selection: {avg_accuracy_svm_before:.4f}")
print("*************************************************************")





```

```python
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score
from imblearn.over_sampling import SMOTE, ADASYN

# Step 1: Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Feature Scaling
scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Step 2: SMOTE
print("********** SMOTE Results **********")
smote = SMOTE(random_state=42)
X_smote, y_smote = smote.fit_resample(X_train, y_train)

# Logistic Regression with SMOTE
model_lr = LogisticRegression()
model_lr.fit(X_smote, y_smote)
y_train_pred = model_lr.predict(X_smote)
y_test_pred = model_lr.predict(X_test)

train_accuracy_lr_smote = accuracy_score(y_smote, y_train_pred)
test_accuracy_lr_smote = accuracy_score(y_test, y_test_pred)
print(f"Logistic Regression - Training Accuracy (SMOTE): {train_accuracy_lr_smote:.4f}")
print(f"Logistic Regression - Testing Accuracy (SMOTE): {test_accuracy_lr_smote:.4f}")

# Cross-validation with Logistic Regression
kfold = KFold(n_splits=10, shuffle=True, random_state=42)
cv_scores_lr_smote = cross_val_score(model_lr, X, y, cv=kfold, scoring='accuracy')
print(f"Logistic Regression - CV Accuracy (SMOTE): {cv_scores_lr_smote.mean():.4f}")

# SVM with SMOTE
model_svm = SVC()
model_svm.fit(X_smote, y_smote)
y_train_pred = model_svm.predict(X_smote)
y_test_pred = model_svm.predict(X_test)

train_accuracy_svm_smote = accuracy_score(y_smote, y_train_pred)
test_accuracy_svm_smote = accuracy_score(y_test, y_test_pred)
print(f"SVM - Training Accuracy (SMOTE): {train_accuracy_svm_smote:.4f}")
print(f"SVM - Testing Accuracy (SMOTE): {test_accuracy_svm_smote:.4f}")

# Cross-validation with SVM
cv_scores_svm_smote = cross_val_score(model_svm, X, y, cv=kfold, scoring='accuracy')
print(f"SVM - CV Accuracy (SMOTE): {cv_scores_svm_smote.mean():.4f}")

# Step 3: ADASYN
print("********** ADASYN Results **********")
adasyn = ADASYN(random_state=42)
X_adasyn, y_adasyn = adasyn.fit_resample(X_train, y_train)

# Logistic Regression with ADASYN
model_lr.fit(X_adasyn, y_adasyn)
y_train_pred = model_lr.predict(X_adasyn)
y_test_pred = model_lr.predict(X_test)

train_accuracy_lr_adasyn = accuracy_score(y_adasyn, y_train_pred)
test_accuracy_lr_adasyn = accuracy_score(y_test, y_test_pred)
print(f"Logistic Regression - Training Accuracy (ADASYN): {train_accuracy_lr_adasyn:.4f}")
print(f"Logistic Regression - Testing Accuracy (ADASYN): {test_accuracy_lr_adasyn:.4f}")

# Cross-validation with Logistic Regression
cv_scores_lr_adasyn = cross_val_score(model_lr, X, y, cv=kfold, scoring='accuracy')
print(f"Logistic Regression - CV Accuracy (ADASYN): {cv_scores_lr_adasyn.mean():.4f}")

# SVM with ADASYN
model_svm.fit(X_adasyn, y_adasyn)
y_train_pred = model_svm.predict(X_adasyn)
y_test_pred = model_svm.predict(X_test)

train_accuracy_svm_adasyn = accuracy_score(y_adasyn, y_train_pred)
test_accuracy_svm_adasyn = accuracy_score(y_test, y_test_pred)
print(f"SVM - Training Accuracy (ADASYN): {train_accuracy_svm_adasyn:.4f}")
print(f"SVM - Testing Accuracy (ADASYN): {test_accuracy_svm_adasyn:.4f}")

# Cross-validation with SVM
cv_scores_svm_adasyn = cross_val_score(model_svm, X, y, cv=kfold, scoring='accuracy')
print(f"SVM - CV Accuracy (ADASYN): {cv_scores_svm_adasyn.mean():.4f}")
```
