```python
import pandas as pd
import numpy as np
from scipy.stats import norm
```

```python
df = pd.read_csv('/content/drive/MyDrive/ML/WA_Fn-UseC_-HR-Employee-Attrition.csv')
```

```python
description = df.describe()

# Print the summary statistics
print(description)
```

```python
# Step 2: Select the column for the Z-test
column_name = "MonthlyIncome"
hypothesized_mean = 6500
```

```python
# Step 3: Extract sample data
sample_data = df[column_name]
```

```python
# Step 4: Compute sample statistics
sample_mean = np.mean(sample_data)         # Sample mean (x̄)
sample_std = np.std(sample_data, ddof=1)   # Sample standard deviation (s)
sample_size = len(sample_data)             # Sample size (n)
```

```python
# Step 5: Compute the standard error (SE)
standard_error = sample_std / np.sqrt(sample_size)
```

```python
# Step 6: Compute the Z-score using the hypothesized mean
z_score = (sample_mean - hypothesized_mean) / standard_error
```

```python
# Step 7: Compute the p-value (two-tailed test)
p_value = 2 * norm.sf(abs(z_score))  # Use survival function for precision
```

```python
# Step 8: Determine critical value at α = 0.05
alpha = 0.05
critical_value = norm.ppf(1 - alpha / 2)  # ±1.96 for two-tailed test
```

```python
# Step 9: Decision making
if abs(z_score) > critical_value:
    decision = "Reject H₀ (The average monthly income is significantly different from $6500)."
else:
    decision = "Fail to reject H₀ (No significant difference from $6500)."
```

```python
# Step 10: Print results
print("\nZ-Test on Monthly Income:")
print(f"Total Employees: {sample_size}")
print(f"Sample Mean: ${sample_mean:.2f}")
print(f"Hypothesized Mean: ${hypothesized_mean}")
print(f"Sample Standard Deviation: ${sample_std:.2f}")
print(f"Standard Error: ${standard_error:.2f}")
print(f"Z-Score: {z_score:.4f}")
print(f"P-Value: {p_value:.15f}")  # Rounded to 15 decimal places
print(f"Critical Value: ±{critical_value:.4f}")
print(f"Decision: {decision}")
```
